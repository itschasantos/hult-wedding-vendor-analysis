#### Introduction to the Problem: Price vs Customer Satisfaction
In today's competitive wedding market, understanding the relationship between price and customer satisfaction for cake functions is crucial for choosing the right vendor partners and maximizing customer happiness. Price, as defined by EconomicsOnline (2020), is the monetary value exchanged for goods or services. In our case, price represents the tiers offered by our partnered vendors. Customer satisfaction, according to ASQ (n.d.), reflects the happiness of customers with a product or service, often measured through surveys and ratings. This analysis, based on past customer ratings from our different vendor partners across all departments, aims to answer: Does prices affect customer satisfaction positively? By understanding this relationship, we can make informed decisions about vendor partnerships, offering customers the best combination of budget-friendliness and satisfaction. 

```python
# importing libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats

# reading the data from excel
file = 'WEDDINGS.xlsx'

# assigning data and naming as project
project = pd.read_excel(io = file)
```

#### Summary and Results:

```python
# instantiating a scatter plot for carat and price
sns.lmplot(x          = 'price_id' , # x-axis feature
           y          = 'avg_ratings' , # y-axis feature
           hue        = None    , # categorical data for subsets
           scatter    = True    , # renders a scatter plot
           fit_reg    = True   , # renders a regression line
           aspect     = 2       , # aspect ratio for plot
           data       = project) # DataFrame where features exist


# formatting and displaying the plot
plt.title       (label  = 'Fig 1: Price and Avg Ratings')
plt.xlabel      (xlabel = 'Price (in degree)'          )
plt.ylabel      (ylabel = 'Avg Ratings'           )
plt.xlim        (left   = 0, right = 4.5          ) # where x will end
plt.tight_layout(pad    = 2                    )
plt.show        (block  = True                    )

# Performing hypothesis testing to test significance of data
proj_price = project.loc[:, 'price_id']
proj_ratings = project.loc[:, 'avg_ratings']

# Convert prices and ratings to numeric data types (floats)
proj_price = proj_price.astype(float)
proj_ratings = proj_ratings.astype(float)

# Calculate the t-statistic and p-value for the correlation
t_statistic, p_value = stats.pearsonr(proj_price, proj_ratings)

# Print the t-statistic and p-value
print(f"""
Summary of Significance Testing via Correlation: 
t-statistic / correlation: {round(t_statistic,4)}
p-value:                   {round(p_value,4)}""")
```

The analysis revealed that positive price changes do not significantly affect overall customer satisfaction as seen from figure 1, which also contains regression line. This could be due to several factors, such as demographic difference in preference where the basis of each customer's rating are vastly different given the intricacies of wedding expectations from families and couples as a whole. Another reason could be the sample size might be insufficient for conclusive testing or that the reviews of the vendors included in the sample size are not significant enough as seen on the results of the testing where correlation or t-value is -0.0267, showing very low negative correlations while p-value is 0.5749.

However, analyzing the relationship between prices and customer ratings per function highlights interesting trends in extreme price ranges. As seen in figure 2, at the minimum and maximum price points, we observe distinct patterns, both positive and negative. This suggests that while price doesn't significantly impact customer satisfaction overall, it can play a nuanced in function-specific role, most notably is cake where a clear postive trend is shown between prices and customer ratings.
